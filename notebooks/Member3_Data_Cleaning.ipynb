# Data Cleaning and Methodology
## Predicting Road Accident Severity Using Data Science

**Team Role:** Member 3 — Data Cleaning, Preprocessing, and Methodology

**Dataset:** UK Road Safety Collisions, 2024–2025

**Objective:** Inspect the dataset, identify missing values and duplicate records, check data types and invalid values, document preprocessing decisions, and export a cleaned dataset for subsequent analysis.

**Tools:** Python, Pandas, and NumPy in Google Colab.

```python

# Cell 1: Load the dataset

import pandas as pd
import numpy as np

DATA_FILE = "/content/collisions_2024_2025_prepared.csv.gz"

df = pd.read_csv(DATA_FILE, low_memory=False)

clean_df = df.copy()

print("Dataset loaded successfully!")
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])

display(clean_df.head())
```

```python

# Cell 2: Check missing values

missing_values = clean_df.isnull().sum()

print("Missing values in each column:")
display(missing_values[missing_values > 0].sort_values(ascending=False))
```

```python

# Cell 3: Check and Remove Exact Duplicates

before = len(clean_df)

clean_df = clean_df.drop_duplicates().copy()

removed = before - len(clean_df)

print("Duplicate rows removed:", removed)
print("Rows remaining:", len(clean_df))
print("Columns remaining:", len(clean_df.columns))
```

```python

# Cell 4: Validate Years and Accident Severity

print("Records per year:")
display(clean_df["collision_year"].value_counts().sort_index())

print("\nAccident severity distribution:")
display(clean_df["collision_severity"].value_counts(dropna=False).sort_index())

print("\nMissing target values:", clean_df["collision_severity"].isna().sum())
```

```python
# Cell 5: Review Missing Values and Data Types

missing_summary = pd.DataFrame({
    "Missing Count": clean_df.isnull().sum(),
    "Missing Percentage": (
        clean_df.isnull().mean() * 100
    ).round(2),
    "Data Type": clean_df.dtypes.astype(str)
})

missing_summary = missing_summary[
    missing_summary["Missing Count"] > 0
].sort_values("Missing Percentage", ascending=False)

print("Columns with missing values:", len(missing_summary))
display(missing_summary)
```

```python
# Cell 6: Inspect Related Columns

columns_to_check = [
    "carriageway_hazards",
    "carriageway_hazards_label",
    "junction_detail",
    "junction_detail_label",
    "junction_group",
    "weather_conditions",
    "weather_conditions_label",
    "weather_group",
    "road_surface_conditions",
    "road_surface_conditions_label",
    "surface_group"
]

for column in columns_to_check:
    if column in clean_df.columns:
        print(f"\n--- {column} ---")
        print("Missing values:", clean_df[column].isnull().sum())
        print("Unique values:", clean_df[column].nunique(dropna=True))
        display(clean_df[column].value_counts(dropna=False).head(8))
```

```python
# Cell 7: Verify Carriageway Hazard Labels

print("Hazard code and label comparison:")

hazard_check = clean_df[
    ["carriageway_hazards", "carriageway_hazards_label"]
].copy()

print("\nMissing values in each column:")
display(hazard_check.isnull().sum())

print("\nRows where the code exists but the label is missing:")
display(
    hazard_check[
        hazard_check["carriageway_hazards"].notna()
        & hazard_check["carriageway_hazards_label"].isna()
    ].head(10)
)

print("\nRows where both values exist:")
display(
    hazard_check[
        hazard_check["carriageway_hazards"].notna()
        & hazard_check["carriageway_hazards_label"].notna()
    ].head(10)
)
```

```python
# Cell 8: Check Missing Hazard Labels by Code

hazard_summary = (
    clean_df.groupby("carriageway_hazards", dropna=False)
    .agg(
        Total_Rows=("carriageway_hazards_label", "size"),
        Missing_Labels=("carriageway_hazards_label", lambda x: x.isna().sum()),
        Available_Labels=("carriageway_hazards_label", lambda x: x.notna().sum())
    )
)

hazard_summary["Missing_Label_Percentage"] = (
    hazard_summary["Missing_Labels"] / hazard_summary["Total_Rows"] * 100
).round(2)

display(hazard_summary.sort_index())
```

```python
# Cell 9: Create a Safe Working Copy

# Keep the original dataset unchanged
original_df = df.copy()

# Create the working dataset for cleaning
clean_df = original_df.copy()

# Remove exact duplicate rows
rows_before = len(clean_df)
clean_df = clean_df.drop_duplicates().copy()
duplicates_removed = rows_before - len(clean_df)

print("Original dataset rows:", len(original_df))
print("Rows after duplicate removal:", len(clean_df))
print("Duplicate rows removed:", duplicates_removed)
print("Columns:", len(clean_df.columns))
print("Missing target values:", clean_df["collision_severity"].isna().sum())
```

```python
# Cell 10: Validate Data Types and Values

# Check the data types of all columns
print("Data types:")
display(clean_df.dtypes.value_counts().rename("Number of Columns"))

# Check the target variable
print("\nAccident severity values:")
display(clean_df["collision_severity"].value_counts().sort_index())

# Check the dataset years
print("\nCollision years:")
display(clean_df["collision_year"].value_counts().sort_index())

# Check invalid geographic coordinates
invalid_latitude = (
    (clean_df["latitude"] < -90) |
    (clean_df["latitude"] > 90)
).sum()

invalid_longitude = (
    (clean_df["longitude"] < -180) |
    (clean_df["longitude"] > 180)
).sum()

print("\nInvalid latitude values:", invalid_latitude)
print("Invalid longitude values:", invalid_longitude)

# Check invalid speed limits
print(
    "\nZero or negative speed limits:",
    (clean_df["speed_limit"] <= 0).sum()
)

# Check unexpected target categories
unexpected_severity = clean_df.loc[
    ~clean_df["collision_severity"].isin([1, 2, 3]),
    "collision_severity"
]

print("\nUnexpected severity values:")
display(unexpected_severity.value_counts(dropna=False))
```

```python
# Cell 11: Identify Label Columns

label_columns = [
    column for column in clean_df.columns
    if column.endswith("_label")
]

print("Number of label columns:", len(label_columns))
print("\nLabel columns:")
for column in label_columns:
    print(column)

print("\nTarget column preserved:",
      "collision_severity" in clean_df.columns)
```

```python
# Cell 12: Remove Unreliable and Redundant Label Columns

# Keep original coded columns for analysis
columns_to_remove = [
    "collision_severity_label",
    "day_of_week_label",
    "police_force_label",
    "first_road_class_label",
    "road_type_label",
    "speed_limit_label",
    "junction_detail_label",
    "junction_control_label",
    "second_road_class_label",
    "pedestrian_crossing_label",
    "light_conditions_label",
    "weather_conditions_label",
    "road_surface_conditions_label",
    "carriageway_hazards_label",
    "urban_or_rural_area_label",
    "did_police_officer_attend_scene_of_accident_label",
    "trunk_road_flag_label",
    "collision_injury_based_label"
]

# Remove only columns that exist
columns_to_remove = [
    column for column in columns_to_remove
    if column in clean_df.columns
]

clean_df = clean_df.drop(columns=columns_to_remove).copy()

print("Label columns removed:", len(columns_to_remove))
print("Rows remaining:", len(clean_df))
print("Columns remaining:", len(clean_df.columns))
print("Target preserved:", "collision_severity" in clean_df.columns)
```

```python
# Cell 13: Handle Missing Values in Categorical Numeric Columns

# Numeric-coded categorical columns
categorical_numeric_columns = [
    "police_force",
    "first_road_class",
    "road_type",
    "speed_limit",
    "junction_detail",
    "junction_control",
    "second_road_class",
    "pedestrian_crossing",
    "light_conditions",
    "weather_conditions",
    "road_surface_conditions",
    "carriageway_hazards",
    "urban_or_rural_area",
    "did_police_officer_attend_scene_of_accident",
    "trunk_road_flag"
]

# Process only columns that exist and contain missing values
for column in categorical_numeric_columns:
    if column in clean_df.columns and clean_df[column].isna().any():
        clean_df[column] = clean_df[column].fillna(-1)

print("Missing values handled in numeric-coded categorical columns.")

remaining_missing = clean_df.isna().sum()
display(
    remaining_missing[remaining_missing > 0]
    .sort_values(ascending=False)
)

print("\nRows remaining:", len(clean_df))
print("Columns remaining:", len(clean_df.columns))
```

```python
# Cell 14: Handle Missing Values in Categorical Text Columns

text_categorical_columns = [
    "junction_group",
    "weather_group",
    "surface_group",
    "light_group",
    "darkness_lighting",
    "area",
    "road_class_group"
]

for column in text_categorical_columns:
    if column in clean_df.columns:
        clean_df[column] = clean_df[column].fillna("Unknown")

print("Missing values handled in categorical text columns.")

print("\nRemaining missing values:")
remaining_missing = clean_df.isna().sum()
display(remaining_missing[remaining_missing > 0])

print("\nRows remaining:", len(clean_df))
print("Columns remaining:", len(clean_df.columns))
```

```python
# Cell 15: Handle Remaining Missing Values

# Keep all accident records.
# Missing coordinates remain missing because we cannot infer their locations.
# Use a separate indicator for unknown high-wind information.
clean_df["high_winds_unknown"] = clean_df["high_winds"].isna().astype(int)

print("Unknown high_winds records:",
      clean_df["high_winds_unknown"].sum())

print("Missing latitude:", clean_df["latitude"].isna().sum())
print("Missing longitude:", clean_df["longitude"].isna().sum())

print("Rows preserved:", len(clean_df))
print("Columns:", len(clean_df.columns))
```

```python
# Cell 16: Validate the Cleaned Dataset

print("FINAL DATASET VALIDATION")
print("-" * 35)

print("Rows:", len(clean_df))
print("Columns:", len(clean_df.columns))
print("Duplicate rows:", clean_df.duplicated().sum())

print("\nMissing values by column:")
remaining = clean_df.isna().sum()
display(remaining[remaining > 0])

print("\nTarget validation:")
print("Missing target values:",
      clean_df["collision_severity"].isna().sum())
print("Target distribution:")
display(clean_df["collision_severity"].value_counts().sort_index())

print("\nYear validation:")
display(clean_df["collision_year"].value_counts().sort_index())

print("\nHigh-wind unknown indicator:")
display(clean_df["high_winds_unknown"].value_counts().sort_index())
```

```python
# Cell 17: Check Potential Invalid Values

print("=== DATA QUALITY CHECK ===")

# Check numerical ranges
print("\n1. Latitude outside valid range:")
print(((clean_df["latitude"] < -90) |
       (clean_df["latitude"] > 90)).sum())

print("\n2. Longitude outside valid range:")
print(((clean_df["longitude"] < -180) |
       (clean_df["longitude"] > 180)).sum())

print("\n3. Zero or negative speed limits:")
print((clean_df["speed_limit"] <= 0).sum())

# Check target values
print("\n4. Collision severity values:")
print(clean_df["collision_severity"].value_counts(dropna=False).sort_index())

# Check categorical codes
columns_to_check = [
    "road_type",
    "junction_detail",
    "junction_control",
    "light_conditions",
    "weather_conditions",
    "road_surface_conditions",
    "pedestrian_crossing",
    "carriageway_hazards"
]

print("\n5. Unique values in categorical columns:")
for col in columns_to_check:
    if col in clean_df.columns:
        print(f"\n{col}:")
        print(sorted(clean_df[col].dropna().unique().tolist()))
```

```python
# Cell 17: Inspect Invalid Speed Limits

invalid_speed = clean_df[
    clean_df["speed_limit"].isna() |
    (clean_df["speed_limit"] <= 0)
]

print("Records with invalid speed limits:", len(invalid_speed))

display(invalid_speed[
    ["collision_year", "speed_limit", "road_type",
     "urban_or_rural_area", "collision_severity"]
])
```

```python
# Cell 19: Verify Missing Speed Limit Values

print("Speed limit values in the original dataset:")
display(df["speed_limit"].value_counts(dropna=False).sort_index())

print("\nSpeed limit values in the cleaned dataset:")
display(clean_df["speed_limit"].value_counts(dropna=False).sort_index())

print("\nRows with speed_limit = -1:")
display(
    clean_df.loc[
        clean_df["speed_limit"] == -1,
        ["collision_year", "speed_limit", "road_type",
         "urban_or_rural_area", "collision_severity"]
    ]
)
```

```python
# Cell 20: Inspect Date and Time Columns

date_columns = [
    col for col in clean_df.columns
    if "date" in col.lower() or "time" in col.lower()
]

print("Date and time columns:", date_columns)

for col in date_columns:
    print(f"\n--- {col} ---")
    print("Data type:", clean_df[col].dtype)
    print("Missing values:", clean_df[col].isna().sum())
    display(clean_df[col].head())
```

```python
# Cell 21: Convert Date and Datetime Columns

# Convert date to datetime format
clean_df["date"] = pd.to_datetime(
    clean_df["date"],
    format="%d/%m/%Y",
    errors="coerce"
)

# Convert datetime to datetime format
clean_df["datetime"] = pd.to_datetime(
    clean_df["datetime"],
    errors="coerce"
)

print("Date dtype:", clean_df["date"].dtype)
print("Datetime dtype:", clean_df["datetime"].dtype)

print("\nMissing dates after conversion:")
print(clean_df["date"].isna().sum())

print("\nMissing datetimes after conversion:")
print(clean_df["datetime"].isna().sum())

display(clean_df[["date", "time", "datetime", "time_band"]].head())
```

```python
# Cell 22: Validate Date and Time Consistency

# Reconstruct datetime from date and time
combined_datetime = pd.to_datetime(
    clean_df["date"].dt.strftime("%Y-%m-%d") + " " +
    clean_df["time"].astype(str),
    errors="coerce"
)

# Compare with the existing datetime column
mismatch = (
    combined_datetime.notna() &
    clean_df["datetime"].ne(combined_datetime)
)

print("Unparseable date/time values:",
      combined_datetime.isna().sum())

print("Datetime mismatches:", mismatch.sum())

if mismatch.any():
    display(clean_df.loc[
        mismatch,
        ["date", "time", "datetime"]
    ].head(10))
else:
    print("Date, time, and datetime are consistent.")
```

```python
# Cell 23: Final Missing Values Review

missing_summary = pd.DataFrame({
    "Missing_Count": clean_df.isna().sum(),
    "Missing_Percentage": (
        clean_df.isna().mean() * 100
    ).round(2)
})

missing_summary = missing_summary[
    missing_summary["Missing_Count"] > 0
].sort_values("Missing_Count", ascending=False)

print("Remaining Missing Values:")
display(missing_summary)

print("\nDataset shape:", clean_df.shape)
print("Duplicate rows:", clean_df.duplicated().sum())
```

```python
# Cell 24: Add Missing-Value Indicators

# Identify records with missing coordinates
clean_df["coordinates_unknown"] = (
    clean_df["latitude"].isna() |
    clean_df["longitude"].isna()
).astype(int)

# Keep original missing values; do not invent coordinates
print("Missing high_winds:", clean_df["high_winds"].isna().sum())
print("Missing latitude:", clean_df["latitude"].isna().sum())
print("Missing longitude:", clean_df["longitude"].isna().sum())

print("\nMissing-coordinate indicator:")
display(clean_df["coordinates_unknown"].value_counts().sort_index())

print("\nDataset shape:", clean_df.shape)
```

```python

# Cell 25: Export Cleaned Dataset

OUTPUT_FILE = "/content/UK_Road_Accidents_Cleaned_2024_2025.csv"

clean_df.to_csv(OUTPUT_FILE, index=False)

# Verify the exported file
check_df = pd.read_csv(OUTPUT_FILE, low_memory=False)

print("Export successful!")
print("Original cleaned shape:", clean_df.shape)
print("Exported shape:", check_df.shape)
print("Duplicate rows:", check_df.duplicated().sum())
print("Target missing values:", check_df["collision_severity"].isna().sum())
print("File path:", OUTPUT_FILE)
```
